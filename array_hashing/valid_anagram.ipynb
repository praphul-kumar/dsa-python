242. Valid Anagram

Given two strings s and t, return true if t is an anagram of s, and false otherwise.

Example 1:

Input:
s = "anagram"
t = "nagaram"

Output: true

Example 2:

Input:
s = "rat"
t = "car"

Output: false


Constraints:

1 <= s.length, t.length <= 5 * 104
s and t consist of lowercase English letters.

```python
def isAnagram(s: str, t: str):
    if len(s) != len(t):
        return False
    
    s_map: dict = {}
    t_map: dict = {}

    for x in s:
        s_map[x] = s_map.get(x, 0) + 1

    for x in t:
        t_map[x] = t_map.get(x, 0) + 1

    return s_map == t_map

# Time: O(m + n) = O(n)
# Space: O(m + n) = O(n) 
```

```python
def isAnagram(s: str, t: str):
    if len(s) != len(t):
        return False
    
    s_map: dict = {}

    for x in s:
        s_map[x] = s_map.get(x, 0) + 1

    for x in t:
        if x not in s_map:
            return False
        
        s_map[x] -= 1

        if s_map[x] < 0:
            return False

    return True

# Time: O(m + n) = O(n)
# Space: O(m) = O(n) 
```

```python
def isAnagram(s: str, t: str):
    if len(s) != len(t):
        return False
    for idx in set(s):
        if s.count(idx) != t.count(idx):
            return False
    return True

# Time: O(n.n) = O(n^2)
# Space: O(n)
```
