## Encode and Decode Strings

Design an algorithm to encode a list of strings to a string. The encoded string is then sent over the network and is decoded back to the original list of strings.

Machine 1 (sender) has the function:

```
String encode(List<String> strs) {
    // ... your code
    return encoded_string;
}
```

Machine 2 (receiver) has the function:

```
List<String> decode(String encoded_string) {
    // ... your code
    return decoded_strs;
}
```

So Machine 1 does:

```
String encoded_string = encode(strs);
```

and Machine 2 does:

```
List<String> decoded_strs = decode(encoded_string);
```
decoded_strs in Machine 2 should be the same as the input strs in Machine 1.

Implement the encode and decode methods.

#### Example 1:

Input: strs = ["Hello","World"]

Output: ["Hello","World"]


#### Explanation:
```
Solution solution = new Solution();
String encoded_string = solution.encode(strs);

// Machine 1 ---encoded_string---> Machine 2

List<String> decoded_strs = solution.decode(encoded_string);
```

#### Example 2:

Input: strs = [""]

Output: [""]

#### Constraints:

0 <= strs.length < 100
0 <= strs[i].length < 200
strs[i] contains any possible characters out of 256 valid ASCII characters.


## ALGORITHM

#### Encoding
    1. If the input list is empty, return an empty string.
    2. Create an empty list to store the sizes of each string.
    3. For each string, append its length to the sizes list.
    4. Build a single string by:
        4.1 Writing all sizes separated by commas.
        4.2 Adding a '#' to mark the end of the size section.
        4.3 Appending all the actual strings in order.
    5. Return the final encoded string.

#### Decoding
    1. If the encoded string is empty, return an empty list.
    2. Read characters from the start until reaching '#' to extract all recorded sizes:
    3. Parse each size by reading until a comma.
    4. After the '#', extract substrings according to the sizes list:
    5. For each size, read that many characters and append the substring to the result.
    6. Return the list of decoded strings.
    

````python
class Encoder:
    _SEPERATOR: str = "#"
    def encode(self, strs: list[str]) -> str:
        if len(strs) <= 0:
            return None

        lens = []
        encoded_str = ""
        for s in strs:
            lens.append(str(len(s)))
            encoded_str += s

        return f"{','.join(lens)}{self._SEPERATOR}{encoded_str}"


    def decode(self, s: str) -> list[str]:
        if not s:
            return []

        sep_idx = s.index(self._SEPERATOR)

        len_parts = s[:sep_idx]
        str_parts = s[sep_idx + 1:]

        start = 0
        end = 0
        res = []
        for len in len_parts.split(","):
            end += int(len)
            print(start, end)
            res.append(str_parts[start:end])
            start = end

        return res
````

````python
# Example 1

input = ["Hello", "World"]

encoder = Encoder()

encoded_str = encoder.encode(input)
print(encoded_str)

decoded_str = encoder.decode(encoded_str)
print(decoded_str)
````
